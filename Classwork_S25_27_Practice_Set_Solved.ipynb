```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

## Section A

```python
"""Q1. Line Plot and Markers
Create a line plot using the following data:
x = [1, 2, 3, 4, 5, 6]
y = [4, 9, 7, 12, 10, 16]
Use a blue dashed line, circular markers of size 8, a title, axis labels, and gridlines. Explain whether the values increase consistently.
"""

x = [1, 2, 3, 4, 5, 6]
y = [4, 9, 7, 12, 10, 16]
plt.plot(x, y, marker = 'o', color = 'blue', linestyle = '--', ms = 8)
plt.title('Line Plot with Markers')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
"""Q2. Multiple Lines in One Graph
Given:
days = [1, 2, 3, 4, 5, 6]
product_A = [12, 15, 18, 17, 22, 25]
product_B = [10, 14, 13, 19, 20, 23]
Plot both products in the same figure using different colours, markers, and line styles. Include a legend.
Identify the days when Product B exceeds Product A.
"""

days = [1, 2, 3, 4, 5, 6]
product_A = [12, 15, 18, 17, 22, 25]
product_B = [10, 14, 13, 19, 20, 23]

plt.plot(days, product_A, marker = 'o', color = 'blue', linestyle = '--', label = 'Product A')
plt.plot(days, product_B, marker = 'o', color = 'red', linestyle = '--', label = 'Product B')
plt.title('Plot for Product A and Product B over 6 days')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.legend()
plt.show()
```

```python
"""Q3. Understanding Subplots
Generate the following values:
x = np.arange(1, 11)
y1 = x**2
y2 = x**3
Create two subplots placed side by side. Display a line plot of y1 in the first subplot and a scatterplot of y2 in the second. Add separate titles and axis labels.
"""
x = np.arange(1, 11)
y1 = x**2
y2 = x**3
plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title('Line Plot of y1')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.subplot(1, 2, 2)
plt.scatter(x, y2)
plt.title('Scatter Plot of y2')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')

plt.tight_layout()
```

```python
"""Q4. Changing Marker Sizes and Transparency
Given:
x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]
sizes = [50, 100, 150, 200, 250]
Create a scatterplot where each point has a different size using sizes. Set transparency to 0.5. Explain how marker size and transparency affect the visualization.
"""
x = [1, 2, 3, 4, 5]
y = [10, 15, 12, 20, 18]
sizes = [50, 100, 150, 200, 250]
plt.scatter(x, y, s = sizes, alpha = 0.5)
plt.title('Scatter Plot with Different Marker Sizes')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

## Section B

```python
"""Q5. Assigning Colours Using a Colormap
Create a scatterplot using:
x = [2, 4, 6, 8, 10]
y = [5, 9, 7, 14, 18]
values = [10, 30, 50, 70, 90]
Assign colors using the viridis colormap based on values. Add a colorbar. Repeat the graph using viridis_r. Explain the difference.
"""
x = [2, 4, 6, 8, 10]
y = [5, 9, 7, 14, 18]
values = [10, 30, 50, 70, 90]
plt.scatter(x, y, c = values, cmap = 'viridis')
plt.colorbar(label = 'Value')
plt.title('Scatter Plot with Colormap')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()

plt.scatter(x, y, c = values, cmap = 'viridis_r')
plt.colorbar(label = 'Value')
plt.title('Scatter Plot with Colormap')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
```

```python
"""Q6. Scatterplot with Three Variables
Given:
unit_cost = [5, 8, 10, 12, 15, 18]
unit_price = [9, 13, 15, 20, 22, 28]
quantity = [100, 80, 120, 60, 90, 40]
Create a scatterplot with Unit Cost on the X-axis and Unit Price on the Y-axis.
Represent Quantity through marker color using the plasma colormap and marker size proportional to Quantity.
Add a colorbar and interpret the relationship.
"""
unit_cost = [5, 8, 10, 12, 15, 18]
unit_price = [9, 13, 15, 20, 22, 28]
quantity = [100, 80, 120, 60, 90, 40]
plt.scatter(unit_cost, unit_price, s = quantity, c = quantity, cmap = 'plasma')
plt.colorbar(label = 'Quantity')
plt.title('Scatter Plot with Three Variables')
plt.xlabel('Unit Cost')
plt.ylabel('Unit Price')
plt.show()
```

```python
"""Q7. Visualizing Different Mathematical Relationships
Generate 100 values of x between 1 and 10 using NumPy. Calculate:
y1 = x
y2 = x**2
y3 = x**3
Create three scatterplots in one figure, one for each relationship.
Use different colours. Compare how the patterns change as the exponent increases.
"""
x = np.linspace(1, 10, 100)
y1 = x
y2 = x**2
y3 = x**3
plt.figure(figsize = (10, 6))
plt.scatter(x, y1, color = 'blue', label = 'Linear')
plt.scatter(x, y2, color = 'red', label = 'Quadratic')
plt.scatter(x, y3, color = 'green', label = 'Cubic')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plots of Different Mathematical Relationships')
plt.legend()
plt.show()
```

```python
"""Q11. Histograms and Bin Sizes
Generate 500 observations from a normal distribution with mean 150 and standard deviation 15, using a fixed random seed 32.
Create three histograms using 5, 15, and 30 bins. Arrange them in separate subplots.
Explain how changing the number of bins affects the appearance and interpretation of the distribution.
"""

np.random.seed(32)
h = np.random.normal(150, 15, 500)
plt.subplot(1, 3, 1)
plt.hist(h, bins = 5)
plt.title('Histogram with 5 bins')
plt.show()

plt.subplot(1, 3, 2)
plt.hist(h, bins = 15)
plt.title('Histogram with 15 bins')
plt.show()

plt.subplot(1, 3, 3)
plt.hist(h, bins = 30)
plt.title('Histogram with 30 bins')
plt.show()
```
